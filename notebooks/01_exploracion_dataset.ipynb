# Exploracion inicial de jupyter-agent-dataset

Fase 0 - analisis rapido por streaming.

```python
from datasets import load_dataset, get_dataset_config_names, load_dataset_builder
repo = 'jupyter-agent/jupyter-agent-dataset'
configs = get_dataset_config_names(repo)
config = configs[0] if configs else None
builder = load_dataset_builder(repo, name=config)
print('config:', config)
print('splits:', builder.info.splits)
print('features:', builder.info.features)
```

```python
stream_ds = load_dataset(repo, name=config, streaming=True)
for split_name, split_ds in stream_ds.items():
    first = next(iter(split_ds))
    print('\n===', split_name, '===')
    print('id:', first.get('id'))
    print('question:', first.get('question'))
    print('packages_used:', first.get('packages_used'))
    print('edu_score:', first.get('edu_score'))
```
